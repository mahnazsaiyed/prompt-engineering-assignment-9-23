# Exercise 3: Self-Reflection Prompt for Improving Output

## Goal
Use AI to create a summary, review its own work, and improve the summary based on specific requirements.

## Tools Used

- Google Colab
- Python
- Google Gemini API

## Text to Summarize

Artificial intelligence is becoming more common in education especially in middle and high schools. Students can use AI tools to explain difficult concepts, brainstorm ideas, summarize information, and receive feedback on their work. These tools can make learning more efficient, but they can also create problems if students depend on them too much. Overusing AI may prevent students from practicing important skills such as critical thinking and problem solving.Students also start to depend on these tools and misuse them leading to bigger problems such as the literacy crisis. Due to this, students should use AI as a learning tool rather than as a replacement for their own thinking. We all need to start using AI as a tool and not something we should be afraid of.

```python
from google import genai
from google.colab import userdata

api_key = userdata.get("GEMINI_API_KEY")
client = genai.Client(api_key=api_key)

print("Gemini API connected successfully!")
```

## Original Summary

For the first attempt, I used a simple prompt asking the AI to summarize the provided text.

```python
original_text = """
Artificial intelligence is becoming more common in education especially in middle and high schools. Students can use AI tools to explain difficult concepts, brainstorm ideas, summarize information, and receive feedback on their work. These tools can make learning more efficient, but they can also create problems if students depend on them too much. Overusing AI may prevent students from practicing important skills such as critical thinking and problem solving.Students also start to depend on these tools and misuse them leading to bigger problems such as the literacy crisis. Due to this, students should use AI as a learning tool rather than as a replacement for their own thinking. We all need to start using AI as a tool and not something we should be afraid of.
"""

original_prompt = f"""
Summarize the following text about AI in education:

{original_text}
"""

response1 = client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents=original_prompt
)

original_summary = response1.text

print("ORIGINAL SUMMARY")
print(original_summary)
```

## Self-Reflection and Critique

Next, I asked the AI to review its original summary using specific requirements and then create an improved version.

```python
reflection_prompt = f"""
You are reviewing a summary that you previously generated.

Original text:
{original_text}

Original summary:
{original_summary}

Critique the original summary using these criteria:

1. Accuracy: Does it correctly represent the main ideas from the original text?
2. Clarity: Is it easy to understand and free of unnecessary wording?
3. Audience: Is it appropriate for a college student?
4. Length: The improved summary should be between 40 and 60 words.
5. Formatting: The improved summary should be one clear paragraph.

First, provide a brief critique identifying what the original summary did well
and what could be improved.

Then create a revised summary that addresses the critique and follows all five
requirements.

Use this format:

CRITIQUE:
[Your critique]

IMPROVED SUMMARY:
[Your revised summary]
"""

response2 = client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents=reflection_prompt
)

improved_output = response2.text

print(improved_output)
```

## Final Reflection

The original summary was accurate, but the self-reflection prompt gave the AI more specific requirements to follow. By checking accuracy, clarity, audience, length, and formatting, the AI was able to review its first response and create a more focused summary.

This showed me that giving AI clear criteria and telling it exactly what you want can help drastically improve the quality of its output. It also makes it easier when you get your output and it is exactly what you are looking for.
